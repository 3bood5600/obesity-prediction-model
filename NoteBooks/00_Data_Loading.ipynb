## 📖 About the Dataset

The Obesity risk dataset contains 20,758 rows and 18 columns.  

### 🏷️ Feature Description:

- **ID**: Unique identifier for each individual  
- **Gender**: The biological sex of the individual (Male/Female)  
- **Age**: The age of the individual in years  
- **Height**: The height of the individual  
- **Weight**: The weight of the individual  
- **family_history_with_overweight**: Family history of overweight (Yes/No)  
- **FAVC**: Frequently consumes high-calorie food (Yes/No)  
- **FCVC**: Frequency of vegetable consumption (1-3 scale)  
- **NCP**: Number of main meals per day  
- **CAEC**: Frequency of eating between meals (Never/Sometimes/Frequently/Always)  
- **SMOKE**: Smoking status (Yes/No)  
- **CH2O**: Daily water intake in liters  
- **SCC**: History of calorie consumption monitoring (Yes/No)  
- **FAF**: Physical activity frequency per week (0-3 scale)  
- **TUE**: Time using electronic devices per day (0-2 scale)  
- **CALC**: Frequency of alcohol consumption (Never/Sometimes/Frequently/Always)  
- **MTRANS**: Main mode of transportation  
- **NObeyesdad**: **TARGET** - Obesity type classification

## 📂 Data Loading
Load the obesity dataset and perform initial inspection.

```python
# Load the dataset
data_path = Path("../data/Obesity_data.csv")

# Check if file exists
if data_path.exists():
    df = pd.read_csv(data_path)
    print(f"✅ Dataset loaded successfully!")
    print(f"📊 Dataset shape: {df.shape}")
    print(f"📁 Columns: {len(df.columns)}")
    print(f"📋 Rows: {len(df)}")
else:
    print(f"❌ Dataset not found at {data_path}")
    print("Please ensure the Obesity_data.csv file is in the data/ directory")
```

## 📊 Basic Data Quality Assessment
Let's check the overall quality of our dataset.

```python
# Missing values analysis
missing_stats = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': (df.isnull().sum() / len(df)) * 100,
    'Data Type': df.dtypes
})

# Filter columns with missing values
missing_stats = missing_stats[missing_stats['Missing Count'] > 0]

if len(missing_stats) > 0:
    print("⚠️ Missing Values Found:")
    display(missing_stats)
else:
    print("✅ No missing values found in the dataset!")
```

```python
# Data quality summary
print("🎯 Data Quality Summary:")
print("=" * 30)
print(f"✅ Total Records: {len(df):,}")
print(f"✅ Total Features: {len(df.columns)}")
print(f"✅ Numerical Features: {len(df.select_dtypes(include=[np.number]).columns)}")
print(f"✅ Categorical Features: {len(df.select_dtypes(include=['object']).columns)}")
print(f"✅ Missing Values: {df.isnull().sum().sum()}")
print(f"✅ Duplicate Rows: {df.duplicated().sum()}")

# Target variable distribution
if 'NObeyesdad' in df.columns:
    print(f"\n🎯 Target Variable (NObeyesdad):")
    print(f"   Unique classes: {df['NObeyesdad'].nunique()}")
    print(f"   Class distribution:")
    target_dist = df['NObeyesdad'].value_counts()
    for class_name, count in target_dist.items():
        percentage = (count / len(df)) * 100
        print(f"     {class_name}: {count:,} ({percentage:.1f}%)")
```

## 💾 Save Processed Data
Export the loaded data for use in subsequent notebooks.

```python
# Create directories if they don't exist
Path("../data").mkdir(exist_ok=True)
Path("../results").mkdir(exist_ok=True)

# Save the loaded data for next stages
output_path = Path("../data/loaded_data.csv")
df.to_csv(output_path, index=False)

print(f"✅ Data saved successfully to: {output_path}")
print(f"📊 Saved {len(df):,} records with {len(df.columns)} features")

# Create a summary file for reference
summary_info = {
    'dataset_shape': list(df.shape),
    'numerical_features': list(df.select_dtypes(include=[np.number]).columns),
    'categorical_features': list(df.select_dtypes(include=['object']).columns),
    'missing_values': int(df.isnull().sum().sum()),
    'duplicate_rows': int(df.duplicated().sum()),
    'target_classes': list(df['NObeyesdad'].unique()) if 'NObeyesdad' in df.columns else []
}

# Save target variable information
if 'NObeyesdad' in df.columns:
    target_col = 'NObeyesdad'
    target_counts = df[target_col].value_counts()
    target_info = {
        'target_column': target_col,
        'target_classes': list(df[target_col].unique()),
        'num_classes': len(df[target_col].unique()),
        'class_distribution': target_counts.to_dict()
    }
    
    import json
    with open('../data/target_classes.json', 'w') as f:
        json.dump(target_info, f, indent=2)
    
    print(f"🎯 Target information saved to ../data/target_classes.json")

print(f"\n📋 Dataset Summary for Next Stages:")
print(f"   Shape: {summary_info['dataset_shape']}")
print(f"   Numerical features: {len(summary_info['numerical_features'])}")
print(f"   Categorical features: {len(summary_info['categorical_features'])}")
print(f"   Target classes: {len(summary_info['target_classes'])}")
```
